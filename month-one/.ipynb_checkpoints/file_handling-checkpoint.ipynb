```python

def create_file(file_name, file_content):
    with open(f"{file_name}","w") as file:
        file.write(file_content)
        print(file_content)



create_file("test.txt", "Apple\nBanana\nCherry")
```

```python
def read_file(file_name):
    with open(f"{file_name}","r") as file:
        lines = list(file.readlines())
        cnt=0
        for line in lines:
            cnt+=1
            print(f"{cnt}: {line.strip()}")



read_file("test.txt")
```

```python

def add_log(log):
    with open("log.txt","a") as file:
        file.write(f"\n{log}")
        print(log)


# add_log("file created successfully at 12:26AM")
add_log("download done successfully at 12:38AM")
```

```python
def dynamic_read_file(file_name,position):
    with open(file_name,"r") as file:
        file.seek(position)
        print(file.read())
        file.seek(0)

dynamic_read_file("log.txt",5)
```

```python
def safe_reader(file_name):

    try:
        with open(file_name, "r") as file:
            data = file.read()
    except:
        print("file doesn't exists")
    else:
        print(data)
    finally:
        print("function task execute")

safe_reader("log.txt")
        
```
