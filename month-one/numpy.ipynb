advance indexing and iterations

```python

print(first_array)
copy_first_array = first_array.copy()
print(copy_first_array)
copy_first_array[3]=100
print(first_array)   # not change main array for use copy()
print(copy_first_array)

print("\n")

print(third_array)
print(third_array[[0,2],[2,2]])
print(first_array[[2,4]],"\n")


for i in third_array:
    for j in i:
        print(j)
    print("\n")
    print(i)

print("\n")

for i in np.nditer(third_array):
    print(i)
```

```python
print(third_array)
one_d_third_array = third_array.ravel(order="F")
print(one_d_third_array)


one_array = [[23,23,34,34,45],[45,56,56,45,34]]
two_array = [[2,4,3,4,5],[6,7,8,4,7]]
print(one_array)
print(two_array)

addition_array = np.concatenate((one_array,two_array),axis=1)
print(addition_array)

print("\n")

print(third_array)
print(third_array.T)   # transpose


print(first_array)
print(np.array_split(first_array,2))    # array split
```

numerical operation

```python
operational_array_one = np.random.randint(1,11,size=(1,5))
operational_array_two = np.random.randint(11,22,size=(1,5))

print(operational_array_one)
print(operational_array_two)

print(operational_array_one + operational_array_two)
print(operational_array_one - operational_array_two)
print(operational_array_one * operational_array_two)
print(operational_array_one / operational_array_two)
print(operational_array_one // operational_array_two)
print(operational_array_one % operational_array_two)


print("\n \n")

# sin, cos, tan, rad2deg, deg2rad,  sqrt(), log10

print(np.sin(operational_array_one))
print(np.rad2deg(operational_array_one))
print(np.deg2rad(operational_array_one))
print(np.sqrt(operational_array_one))
print(np.log10(operational_array_one))


# use filter using all() , any()

print(np.all(operational_array_one > 7))
print(np.any(operational_array_one > 7))
```

shorting

```python
sort_array = [[34,2,34,54],[4,54,23,43]]
print(sort_array)
print("\n")
print(np.sort(sort_array,axis=0))
print(np.sort(sort_array,axis=1))
```

searching and counting

```python
searching_array = np.array(sort_array)
print(searching_array)
print("\n")

print(np.where(searching_array > 10))
print(np.where(searching_array > 10, searching_array,0))


print(np.count_nonzero(searching_array > 5))
print(np.unique(searching_array, return_counts=True))
```
