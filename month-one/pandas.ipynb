```python
import pandas as pd
```

```python
patient_data = pd.read_csv("RedioLens.csv")
print(patient_data.head(2))
print(patient_data.tail(2))
print(patient_data.columns)
print(patient_data.info)
print(patient_data.sample(10))

print(type(patient_data["Age"]))

print(patient_data.describe())
```

python buildin data type to pandas dataframe covert

```python
list_data = [["name","Ruhul Amin"],["age",24],["salary",28000]]
list_df = pd.DataFrame(list_data)
print(list_df.head())


tuple_data = (("name","Ruhul Amin"),("age",24),("salary",28000))
tuple_df = pd.DataFrame(tuple_data)
print(tuple_df.head())

dict_data = {
    "name":["Ruhul Amin","Arafat"],
    "age":[24,24],
    "salary":[28000,25000]
}
dict_df = pd.DataFrame(dict_data)
print(dict_df.head())
```

data accessing and indexing

```python
minimize_df = patient_data
# print(minimize_df.loc[[2,4]])
# print(minimize_df.loc[2:5])

print("\n \n \n")
minimize_col_df = minimize_df.loc[:,["MRN","Patient Name","Sex","Age","Modality","Img"]]
print(minimize_col_df)

print(minimize_col_df.loc[50:100,"Patient Name"])


minimize_idx_df = minimize_col_df.set_index("Patient Name")
print(minimize_idx_df)


print(minimize_idx_df.iloc[3:10])
print(minimize_idx_df.iloc[5:10,1:4])

print(minimize_idx_df.rename(columns={"Sex":"Gender"}))
```

data modification

```python

to_much_minimize = minimize_col_df.head(5)

print(to_much_minimize)

to_much_minimize.drop(0,inplace=True)
print(to_much_minimize)
to_much_minimize.drop("Img",axis=1, inplace=True)
print(to_much_minimize)
to_much_minimize.loc[1,"Modality"]="CR"
print(to_much_minimize)
```

sorting values

```python

sorting_df = minimize_col_df.head(10)
print(sorting_df)
sorting_df["Age"] = sorting_df["Age"].where(
    sorting_df["Age"] != 82.0,
    45
)
print(sorting_df.sort_values(["Age","Img"],ascending=[1,0]))
```

```python
filter_data = sorting_df.loc[(sorting_df["Age"] > 50) & (sorting_df["Img"] < 1000)]
print(filter_data)
```
